```python
import pandas as pd

df = pd.read_csv('../data/processed/bixi_2025_cleaned.csv')

station_summary = df.groupby([
    'STARTSTATIONNAME', 
    'STARTSTATIONARRONDISSEMENT', 
    'STARTSTATIONLATITUDE', 
    'STARTSTATIONLONGITUDE'
]).agg(
    total_departures=('duration_sec', 'count'),
    avg_duration_min=('duration_sec', lambda x: (x.mean() / 60)),
    round_trips=('is_round_trip', 'sum')
).reset_index()

station_summary.to_csv('../data/processed/tableau_station_summary.csv', index=False)
print("Saved: tableau_station_summary.csv")

hourly_day_summary = df.groupby(['day_of_week', 'start_hour']).agg(
    total_trips=('duration_sec', 'count'),
    avg_duration_min=('duration_sec', lambda x: (x.mean() / 60))
).reset_index()

hourly_day_summary.to_csv('../data/processed/tableau_hourly_summary.csv', index=False)
print("Saved: tableau_hourly_summary.csv")

df_sample = df.sample(frac=0.10, random_state=404)
df_sample.to_csv('../data/processed/tableau_bixi_2025_sample10pct.csv', index=False)
print("Saved: tableau_bixi_2025_sample10pct.csv (~1.4M rows)")
```
